```python
def check_even_list(num_list):
    # return all the even numbers in a list
    
    # placeholder variables
    even_numbers = []
    
    for num in num_list:
        if num%2 == 0:
            even_numbers.append(num)
        else:
            pass
    return even_numbers
```

```python
# return the sum of the numbers in the array, ignoring sections of numbers starting with a 6 and extending to the next 9
def summer_69(arr):
    total = 0
    add = True
    for num in arr:
        while add:
            if num != 6:
                total += num
                break
            else:
                add = False
        while not add:
            if num != 9:
                break
            else:
                add = True
                break
    return total
```

```python
# need checks in place to ensure using is inputting a number
# need another check to ensure user is inputting a value within the range of 0-10 
    # (not including 10 since there's only 9 squares in tic tac toe)
def user_choice():
    
    choice = 'wrong' # assign an arbitrary starting placeholder
    acceptable_range = range(0,10)
    within_range = False
    
    while choice.isdigit() == False or within_range == False:
        
        choice = input('Please enter a number (0-10): ')
        
        if choice.isdigit() == False:
            print('Sorry that is not a digit')
        if choice.isdigit() == True:
            if int(choice) in acceptable_range:
                within_range = True
            else:
                print('Sorry, you are out of acceptable range (0-10)')
                within_range = False
    
    return int(choice)
```

```python
# create the tic tac toe board
from IPython.display import clear_output
# clear the output so the board is not re-printed every time

def display_board(board):
    clear_output()
    
    print('   |   |')
    print(' ' + board[7] + ' | ' + board[8] + ' | ' + board[9])
    print('   |   |')
    print('-----------')
    print('   |   |')
    print(' ' + board[4] + ' | ' + board[5] + ' | ' + board[6])
    print('   |   |')
    print('-----------')
    print('   |   |')
    print(' ' + board[1] + ' | ' + board[2] + ' | ' + board[3])
    print('   |   |')
```

```python
# assign X and O to player 1 and player 2
def player_input():
    '''
    OUTPUT = (Player 1 marker, Player 2 marker)
    '''
    
    marker = ''
    
    while marker != 'X' and marker != 'O':
        marker = input('Player 1: Choose X or O: ').upper()
        
    if marker == 'X':
        return ('X','O')
    else:
        return ('O','X')
```

```python
# check all rows, columns, and diagonals to see if a player has won
def win_check(board,mark):
    
    return ((board[7] == mark and board[8] == mark and board[9] == mark) or # across the top
    (board[4] == mark and board[5] == mark and board[6] == mark) or # across the middle
    (board[1] == mark and board[2] == mark and board[3] == mark) or # across the bottom
    (board[7] == mark and board[4] == mark and board[1] == mark) or # down the middle
    (board[8] == mark and board[5] == mark and board[2] == mark) or # down the middle
    (board[9] == mark and board[6] == mark and board[3] == mark) or # down the right side
    (board[7] == mark and board[5] == mark and board[3] == mark) or # diagonal
    (board[9] == mark and board[5] == mark and board[1] == mark)) # diagonal
```

```python
# decide which player goes first
import random

def choose_first():
    
    flip = random.randint(0,1)
    
    if flip == 0:
        return 'Player 1'
    else:
        return 'Player 2'
```

```python
# check if a space on the board is empty
def space_check(board,position):
        
    return board[position] == ' '
```

```python
# check if the board if full
def full_board_check(board):
    
    for i in range(1,10):
        if space_check(board,i): # if position is empty return False (board is not full)
            return False
    return True
```

```python
# ask player for next position and check if that position is free
def player_choice(board):
    
    position = 0
    
    # while the player hasn't chosen a position between 1-9 or if the position is not full
    while position not in [1,2,3,4,5,6,7,8,9] or not space_check(board,position):
        position = int(input('Choose a position (1-9)'))
        
    return position
```

```python
# ask the player if they want to play again
# assume their answer is Yes
def replay():
    choice = input('Play again? Enter Yes or No')
    return choice == 'Yes'
```

```python
print('Welcome to tic tac toe')

while True:
    the_board = [' ']*10
    player1_marker, player2_marker = player_input()
    
    turn = choose_first()
    print(turn + ' will go first')
    
    play_game = input('Ready to play? y or n')
    
    if play_game == 'y':
        game_on = True
    else:
        game_on = False
        
    while game_on:
        
        # Player 1's turn
        if turn == 'Player 1':
            display_board(the_board)
            position = player_choice(the_board)
            place_marker(the_board,player1_marker,position)
            
            # check if Player 1 has won
            if win_check(the_board,player1_marker):
                display_board(the_board)
                print('Player 1 has won!')
                game_on = False
                
            # if Player 1 has not won, check if there is a tie
            else:
                if full_board_check(the_board):
                    display_board(the_board)
                    print('tie game!')
                    game_on = False
                else:
                    turn = 'Player 2'
                    
        # Player 2's turn
        else:
            display_board(the_board)
            position = player_choice(the_board)
            place_marker(the_board,player2_marker,position)
            
            if win_check(the_board,player2_marker):
                display_board(the_board)
                print('Player 2 has won!')
                game_on = False
            else:
                if full_board_check(the_board):
                    display_board(the_board)
                    print('tie game!')
                    game_on = False
                else:
                    turn = 'Player 1'
    
    if not replay():
        break
```

```python
# return the slope and distance of a line with tuples as coordinates
class Line():
    
    def __init__(self,coor1,coor2):
        self.coor1 = coor1
        self.coor2 = coor2
        
    def distance(self):
        x1,y1 = self.coor1
        x2,y2 = self.coor2
        return ((x2-x1)**2 + (y2-y1)**2)**0.5
        
    def slope(self):
        x1,y1 = self.coor1
        x2,y2 = self.coor2
        return (y2-y1)/(x2-x1)
```

```python
# return the volume and surface area of a cylinder
class Cylinder():
    
    pi = 3.14
    
    def __init__(self,height=1,radius=1):
        self.height = height
        self.radius = radius
        
    def volume(self):
        return self.height*Cylinder.pi*(self.radius)**2
    
    def surface_area(self):
        top = 3.14*(self.radius)**2
        return (2*top) + (2*Cylinder.pi*self.radius*self.height)
```

```python
# create a deck of all 52 cards
class Deck:
    
    def __init__(self):
        
        self.all_cards = []
        
        for suit in suits:
            for rank in ranks:
                
                created_card = Card(suit,rank)
                
                self.all_cards.append(created_card)
                
    def shuffle(self):
        # doesn't actually return anything, just shuffles in place
        random.shuffle(self.all_cards)
    
    # return a single card from the deck
    def deal_one(self):
        return self.all_cards.pop()
```

```python
# assume players remove one card from the top of the deck and add one card to the bottom of the deck
# assume top to bottom = left to right in Python
# pop(0) to remove from top (left); append to add to bottom (right)
# use extend() to add multiple cards to the deck; can't use append because the result will be a nested list ['b','c',['d','e']]

class Player:
    
    def __init__(self,name):
        
        self.name = name
        self.all_cards = []
        
    def remove_one(self):
        return self.all_cards.pop(0)
    
    def add_cards(self,new_cards):
        # if multiple cards are being added ie. a list of cards
        if type(new_cards) == type([]):
            self.all_cards.extend(new_cards)
        # for a single card
        else:
            self.all_cards.append(new_cards)
            
    def __str__(self):
        return f'Player {self.name} has {len(self.all_cards)} cards'
```

```python
# assume at least 5 cards per player are required for war

game_on = True
round_num = 0

while game_on:
    
    # state which round it's on
    round_num += 1
    print(f'Round {round_num}')
    
    if len(player_one.all_cards) == 0:
        print('Player One out of cards! Player Two wins!')
        game_on = False
        break
        
    if len(player_two.all_cards) == 0:
        print('Player Two out of cards! Player One wins!')
        game_on = False
        break
    
    # start a new round
    # cards in play
    player_one_cards = []
    # remove a single card
    player_one_cards.append(player_one.remove_one())
    
    player_two_cards = []
    player_two_cards.append(player_two.remove_one())
    
    # while at war
    at_war = True
    
    while at_war:
        
        # using [-1] because new cards are being appended throughout the game
        # if we use [0], the card at [0] will always be the same
        
        if player_one_cards[-1].value > player_two_cards[-1].value:
            
            # player one gets all their cards back and all of player two's cards
            player_one.add_cards(player_one_cards)
            player_one.add_cards(player_two_cards)
            
            at_war = False
        
        elif player_one_cards[-1].value < player_two_cards[-1].value:
            
            # player two gets all their cards back and all of player one's cards
            player_two.add_cards(player_one_cards)
            player_two.add_cards(player_two_cards)
            
            at_war = False
            
        else:
            print('WAR!')
            
            if len(player_one.all_cards) < 5:
                print('Player One unable to declare war')
                print('Player Two wins!')
                # breaks the game_on while loop
                game_on = False
                # breaks the at_war while loop
                break
                
            elif len(player_two.all_cards) < 5:
                print('Player Two unable to declare war')
                print('Player One wins!')
                game_on = False
                break
                
            else:
                for num in range(5):
                    player_one_cards.append(player_one.remove_one())
                    player_two_cards.append(player_two.remove_one())                           
```

```python
class Deck():
    
    def __init__(self):
        # create a deck of 52 cards
        self.deck = []
        for suit in suits:
            for rank in ranks:
                self.deck.append(Card(suit,rank))
            
    def __str__(self):
        # deck composition
        deck_comp = ''
        for card in self.deck:
            deck_comp += '\n' + card.__str__()
        return 'The deck has: ' + deck_comp
    
    def shuffle(self):
        random.shuffle(self.deck)
        
    def deal(self):
        single_card = self.deck.pop()
        return single_card
```

```python
# specify which cards are in a player's hand
class Hand():
    
    def __init__(self):
        self.cards = []
        self.value = 0
        self.aces = 0 # aces can be treated as a 1 or 11
        
    def add_card(self,card):
        # from Deck.deal()
        self.cards.append(card)
        # returns card's value from values dictionary
        self.value += values[card.rank]
        
        # track aces
        if card.rank == 'Ace':
            self.aces += 1
    
    def adjust_for_ace(self):
        
        # if total value > 21 and I still have an ace then change my ace to be a 1 instead of 11
        # self.aces = 0: value of 0 is treated as False; see code below: returns nothing because 0 is False
        while self.value > 21 and self.aces:
            self.value -= 10
            self.aces -= 1
```

```python
class Chips():
    
    # ie. start off with 100 chips
    def __init__(self,total=100):
        self.total = total
        self.bet = 0
        
    def win_bet(self):
        self.total += self.bet
        
    def lose_bet(self):
        self.total -= self.bet
```

```python
def take_bet(chips):
    
    while True:
        
        try:
            chips.bet = int(input('How many chips would you like to bet?'))
        except:
            print('Sorry please provide an integer')
        else:
            if chips.bet > chips.total:
                print(f'Sorry you do not have enough chips! You have: {chips.total}')
            else:
                break
```

```python
def hit(deck,hand):
    
    single_card = deck.deal()
    hand.add_card(single_card)
    hand.adjust_for_ace()
```

```python
def hit_or_stand(deck,hand):
    global playing
    
    while True:
        x = input('Hit or Stand? Enter h or s')
        
        # in case the user accidentally types ie. hit, hh, etc.
        if x[0].lower() == 'h':
            hit(deck,hand)
            
        elif x[0].lower() == 's':
            print("Player stands, dealer's turn")
            playing = False
        
        else:
            print('Sorry I did not understand that, please enter a h or s only')
            continue
        break
```

```python
def show_some(player,dealer):
    
    # show only 1 of 2 of the dealer's cards (1 face up, 1 face down)
    print("\nDealer's Hand:")
    print('First card hidden!')
    
    # show the 2nd card not the first
    print(dealer.cards[1])
    
    # show all 2 cards of the player's hand
    print("\nPlayer's Hand:")
    for card in player.cards:
        print(card)
    
    
def show_all(player,dealer):
    # show all the dealer's cards
    print("\nDealer's Hand:")
    for card in dealer.cards:
        print(card)
    
    # can also use * to print all items in a list; sep = separator
    # this is an alternative to the code directly above
    # see example code below
    # print("\n Dealer's hand: ",*dealer.cards,sep="\n")
    
    # calculate and display value (ie. J+K = 20)
    print(f"Value of Dealer's hand is: {dealer.value}")
        
    # show all the player's cards
    print("\nPlayer's Hand:")
    for card in player.cards:
        print(card)
    print(f"Value of Player's hand is: {player.value}")
```

```python
def player_busts(player,dealer,chips):
    print('BUST Player!')
    chips.lose_bet()
    
def player_wins(player,dealer,chips):
    print('Player wins!')
    chips.win_bet()
    
def dealer_busts(player,dealer,chips):
    print('Player wins! Dealer busted!')
    chips.win_bet()
            
def dealer_wins(player,dealer,chips):
    print('Dealer wins!')
    chips.lose_bet()
    
# both dealer and player have 21: tie    
def push(player,dealer,chips):
    print('Dealer and player tie: PUSH')
```

```python
while True:
    
    print('Welcome to BLACKJACK')
    
    # create and shuffle deck, deal 2 cards to each player
    deck = Deck()
    deck.shuffle()
    
    player_hand = Hand()
    player_hand.add_card(deck.deal())
    player_hand.add_card(deck.deal())
    
    dealer_hand = Hand()
    dealer_hand.add_card(deck.deal())
    dealer_hand.add_card(deck.deal())
    
    # set up the player's chips
    player_chips = Chips()
    
    # player places bet
    take_bet(player_chips)
    
    # show cards but keep one dealer card hidden
    show_some(player_hand,dealer_hand)
    
    while playing:
        
        hit_or_stand(deck,player_hand)
        
        show_some(player_hand,dealer_hand)
        
        if player_hand.value > 21:
            player_busts(player_hand,dealer_hand,player_chips)
            
            break
            
    # if player hasn't busted
    if player_hand.value <= 21:
        
        # in Blackjack, the dealer only hits until they reach 17
        while dealer_hand.value < 17:
            hit(deck,dealer_hand)
            
        show_all(player_hand,dealer_hand)
        
        if dealer_hand.value > 21:
            dealer_busts(player_hand,dealer_hand,player_chips)
        elif dealer_hand.value > player_hand.value:
            dealer_wins(player_hand,dealer_hand,player_chips)
        elif dealer_hand.value < player_hand.value:
            player_wins(player_hand,dealer_hand,player_chips)
        else:
            push(player_hand,dealer_hand)
            
        # inform player of their chips total
        print(f'\nPlayer total chips are at: {player_chips.total}')
        
        # ask player to play again
        new_game = input('Would you like to play another hand? y/n')
        
        if new_game[0].lower() == 'y':
            playing = True
            continue
        else:
            print('Thank you for playing!')
            break
```

```python
def new_decorator(original_func):
    
    # extra functionality to be added to original function
    def wrap_func():
        
        print('Some extra code, before the original function')
        
        original_func()
        
        print('Some extra code, after the original function')
        
    return wrap_func
```

Generators

```python
# a generator returns values one at a time instead of everything at once
# range() is an example of a generator
# ie. range(0,100) doesn't store all values from 0-100 in memory, it only returns the values when they're called

def create_cubes(n):
    result = []
    for x in range(n):
        result.append(x**3)
    return result
```

```python
create_cubes(10)
```

```python
for x in create_cubes(10):
    print(x)
```

```python
# don't need to store results in result = []
# returns the same result but more memory efficient

def create_cubes(n):

    for x in range(n):
        yield x**3
```

```python
for x in create_cubes(10):
    print(x)
```

```python
create_cubes(10)
```

```python
def gen_fibonacci(n):
    
    a = 1
    b = 1
    
    for i in range(n):
        yield a
        a,b = b,a+b
```

```python
for num in gen_fibonacci(10):
    print(num)
```

```python
# go through all pages to get all authors from the website without knowing how many pages there are
url = 'https://quotes.toscrape.com/page/'
```

```python
# assuming you know there are 10 pages on this site

authors = set()

for page in range(1,10):
    
    page_url = url+str(page)
    
    res = requests.get(page_url)
    
    soup = bs4.BeautifulSoup(res.text,'lxml')
    
    for name in soup.select('.author'):
        authors.add(name.text)
```

```python
authors
```

```python
# if you enter a ridiculous page number the page will show 'No quotes found!'
res = requests.get('https://quotes.toscrape.com/page/999999999999/')
```

```python
res.text
```

```python
'No quotes found!' in res.text
```

```python
# assuming you don't know how many pages there are on this site

page_still_valid = True
authors = set()
page = 1

while page_still_valid:
    
    page_url = url+str(page)
    
    res = requests.get(page_url)
    
    if 'No quotes found!' in res.text:
        break # or page_still_valid = False
        
    soup = bs4.BeautifulSoup(res.text,'lxml')
    
    for name in soup.select('.author'):
        authors.add(name.text)
    
    page = page+1
```
